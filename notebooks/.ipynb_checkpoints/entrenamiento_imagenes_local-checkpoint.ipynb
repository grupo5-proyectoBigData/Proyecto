```python
import os
import kagglehub

# 2. Descarga el dataset público
ruta_dataset = kagglehub.dataset_download('davidcodon/skin-diseases-unique-images')
print("¡Dataset descargado con éxito!")
print("Los archivos están guardados en la ruta:", ruta_dataset)
```

```python
from pathlib import Path
import shutil
ruta_dataset = "../data/processed/"
ORIGEN = Path(ruta_dataset)
# Crear la carpeta de destino
DESTINO = Path("/content/skin-diseases")

# Copiar únicamente si todavía no existe
if not DESTINO.exists():
    shutil.copytree(ORIGEN, DESTINO)

DATA_DIR = DESTINO

print("Dataset disponible en:", DATA_DIR)
```

```python
#Comprobamos el espacio libre
total, usado, libre = shutil.disk_usage("/content")

print(f"Espacio total: {total / 1024**3:.2f} GB")
print(f"Espacio usado: {usado / 1024**3:.2f} GB")
print(f"Espacio libre: {libre / 1024**3:.2f} GB")
```

```python
import tensorflow as tf

# Nuevas rutas locales
ruta_train = "/content/skin-diseases/train"
ruta_val = "/content/skin-diseases/val"
ruta_test = "/content/skin-diseases/test"

BATCH_SIZE = 16  # Mantenemos 16, a 32 nos quedamos si memoria
IMG_SIZE = (224, 224)
SEED = 42

# Cargar TRAIN (Se queda listo para el entrenamiento)
train_ds = tf.keras.utils.image_dataset_from_directory(
    ruta_train,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    label_mode='int',
    shuffle=True,
    seed=SEED
)

#Usamos class_names en validación y test para segurar que se usan los mismos índices
class_names = train_ds.class_names
NUM_CLASSES = len(class_names)

# Cargar VAL (Solo se llamará al final de cada época)
val_ds = tf.keras.utils.image_dataset_from_directory(
    ruta_val,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    label_mode="int",
    class_names=class_names,
    shuffle=False
)

# Cargar TEST (Solo ocupará memoria al final de todo el entrenamiento)
test_ds = tf.keras.utils.image_dataset_from_directory(
    ruta_test,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    label_mode="int",
    class_names=class_names,
    shuffle=False
)

# Optimización de flujo (SIN .cache(), solo prefetch)
AUTOTUNE = tf.data.AUTOTUNE
train_ds = train_ds.prefetch(AUTOTUNE)
val_ds = val_ds.prefetch(AUTOTUNE)
test_ds = test_ds.prefetch(AUTOTUNE)

print("¡Datasets configurados para lectura bajo demanda desde el disco duro!")
```

```python
#Comprobamos con algunas imágenes que se han cargado bien
import matplotlib.pyplot as plt
plt.figure(figsize=(12, 8))
for images, labels in train_ds.take(1):
    for i in range(min(12, len(images))):
        ax = plt.subplot(3, 4, i + 1)
        plt.imshow(images[i].numpy().astype("uint8"))
        plt.title(class_names[int(labels[i])])
        plt.axis("off")
    plt.tight_layout()
    plt.show()
```

Durante el entrenamiento generamos variaciones de las imágenes. Las ventajas son:
- Reduce el sobreajuste (overfitting)
- Mejora la capacidad de generalización del modelo
- Permite obtener más variedad de entrenamiento sin aumentar el tamaño del dataset almacenado

```python
#DataAugmentation
from tensorflow.keras import layers, Sequential

data_augmentation = Sequential([
    layers.RandomFlip("horizontal"),
    layers.RandomRotation(0.04),
    layers.RandomZoom(0.10),
    layers.RandomTranslation(
        height_factor=0.05,
        width_factor=0.05
    ),
], name="data_augmentation")
```

```python

from tensorflow.keras import layers, Model
from tensorflow.keras.applications import EfficientNetB0

base_model = EfficientNetB0(
    include_top=False,
    weights="imagenet",
    input_shape=(224, 224, 3)
)

#Congelamos la red preentrenada
base_model.trainable = False

inputs = layers.Input(shape=(224, 224, 3))

# Aumento de datos
x = data_augmentation(inputs)

# Extracción de características
x = base_model(x, training=False)

# Clasificador
x = layers.GlobalAveragePooling2D()(x)
x = layers.Dropout(0.3)(x)

outputs = layers.Dense(
    NUM_CLASSES,
    activation="softmax"
)(x)

model = Model(inputs, outputs)

model = Model(inputs, outputs, name="baseline_efficientnetb0")

model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)
```

```python
import os
import time
from tensorflow.keras.callbacks import (
    EarlyStopping,
    ModelCheckpoint,
    ReduceLROnPlateau,
    CSVLogger
)

RESULTS_DIR = "/content/resultados"
os.makedirs(RESULTS_DIR, exist_ok=True)

callbacks = [
    ModelCheckpoint(
        f"{RESULTS_DIR}/baseline_efficientnetb0.keras",
        monitor="val_loss",
        save_best_only=True,
        verbose=1
    ),

    EarlyStopping(
        monitor="val_loss",
        patience=4,
        restore_best_weights=True,
        verbose=1
    ),

    ReduceLROnPlateau(
        monitor="val_loss",
        factor=0.2,
        patience=2,
        min_lr=1e-7,
        verbose=1
    ),

    CSVLogger(
        f"{RESULTS_DIR}/baseline_history.csv"
    )
]

inicio = time.time()

history = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=15,
    callbacks=callbacks
)

duracion = time.time() - inicio

print(f"Tiempo de entrenamiento: {duracion / 60:.2f} minutos")
```

Representamos la evolución de la precisión y de la pérdida durante el entrenamiento:
- Si entrenamiento y validación mejoran juntos, el modelo está aprendiendo.
- Si mejora el entrenamiento y empeora la validación hay *overfitting*.
- Si los dos bajan puede haber *underfitting*.
- Si la validación mejora más lento, hay que revisar la calidad y el equilibrio de los datos.

```python
def plot_history(history):
    hist = history.history

    fig, axes = plt.subplots(1, 2, figsize=(14, 5))

    axes[0].plot(hist["accuracy"], label="Train")
    axes[0].plot(hist["val_accuracy"], label="Validation")
    axes[0].set_title("Accuracy")
    axes[0].set_xlabel("Época")
    axes[0].legend()

    axes[1].plot(hist["loss"], label="Train")
    axes[1].plot(hist["val_loss"], label="Validation")
    axes[1].set_title("Loss")
    axes[1].set_xlabel("Época")
    axes[1].legend()

    plt.show()

plot_history(history)
```

**Evaluación del modelo**

Una precisión global elevada puede ocultar que el modelo clasifica mal algunas enfermedades menos representadas. Para visualizarlo usaremos:
- *Accuracy*: porcentaje global de aciertos.
- *Precision*: proporción de predicciones positivas correctas.
- *Recall*: proporción de imágenes de una clase que consigue detectar.
- *F1-score*: equilibrio entre precision y recall.
- *Macro F1*: media del F1 de todas las clases, dando el mismo peso a cada enfermedad.

```python
import numpy as np
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay,
    accuracy_score,
    f1_score,
    balanced_accuracy_score
)

# Cargar el mejor modelo guardado
best_model = tf.keras.models.load_model(
    f"{RESULTS_DIR}/baseline_efficientnetb0.keras"
)

# Predicciones sobre validación
y_true = np.concatenate([
    labels.numpy() for images, labels in val_ds
])

y_prob = best_model.predict(val_ds)
y_pred = np.argmax(y_prob, axis=1)

print(classification_report(
    y_true,
    y_pred,
    target_names=class_names,
    digits=4,
    zero_division=0
))

print("Accuracy:", accuracy_score(y_true, y_pred))
print("Macro F1:", f1_score(y_true, y_pred, average="macro"))
print("Balanced accuracy:", balanced_accuracy_score(y_true, y_pred))
```

```python
#Matriz de confusión
cm = confusion_matrix(y_true, y_pred)

fig, ax = plt.subplots(figsize=(12, 12))

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=class_names
)

disp.plot(
    ax=ax,
    xticks_rotation=90,
    cmap="Blues",
    colorbar=False
)

plt.title("Baseline EfficientNetB0 - Validación")
plt.tight_layout()
plt.show()
```

Creamos un registro común para guardar los resultados y comparar los experimentos. Este registro crecerá a medida que incorporemos otros modelos

```python
import pandas as pd
resultado_baseline = {
    "modelo": "EfficientNetB0",
    "tipo": "Transfer learning - capas congeladas",
    "resolucion": "224x224",
    "batch_size": BATCH_SIZE,
    "learning_rate": 0.001,
    "epochs": len(history.history["loss"]),
    "tiempo_minutos": round(duracion / 60, 2),
    "accuracy": float(accuracy_score(y_true, y_pred)),
    "macro_f1": float(f1_score(y_true, y_pred, average="macro")),
    "balanced_accuracy": float(
        balanced_accuracy_score(y_true, y_pred)
    )
}

resultados = pd.DataFrame([resultado_baseline])

resultados.to_csv(
    f"{RESULTS_DIR}/comparativa_modelos.csv",
    index=False
)
```
