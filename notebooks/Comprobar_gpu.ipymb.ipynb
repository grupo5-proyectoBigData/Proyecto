```python
import tensorflow as tf

# 1. Listar los dispositivos físicos disponibles
print("Dispositivos físicos detectados:")
dispositivos = tf.config.list_physical_devices()
for dispositivo in dispositivos:
    print("-", dispositivo)

# 2. Comprobación directa de la GPU
gpu_disponibles = tf.config.list_physical_devices('GPU')
if gpu_disponibles:
    print(f"\n¡Éxito! TensorFlow está usando la GPU: {gpu_disponibles}")
else:
    print("\nTensorFlow NO está detectando ninguna GPU. Está ejecutándose en la CPU.")
```
